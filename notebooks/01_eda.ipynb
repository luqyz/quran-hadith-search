```python
from pathlib import Path
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "data" / "processed").exists())
```

```python
import glob
import pandas as pd

res = pd.read_csv(sorted(glob.glob(str(ROOT / "eval" / "results" / "*.csv")))[-1])
syn = pd.read_csv(ROOT / "eval" / "synthetic_queries.csv", encoding="utf-8-sig")
res = res.merge(syn[["query", "lang"]], on="query", how="left")
res["lang"] = res["lang"].fillna("manual")

# 1. Adakah bahasa soalan memberi kesan?
display(res[res.type != "negative"]
        .groupby(["type", "lang", "method"])["recall@10"].mean().unstack().round(3))

# 2. Berapa soalan positif yang akan tertolak jika threshold BM25 = 14?
pos_bm25 = res[(res.method == "bm25") & (res.type != "negative")]
print(f"Positif di bawah 14: {(pos_bm25.top_score < 14).mean():.1%}")
```

```python
import glob
files = sorted(glob.glob(str(ROOT / "eval" / "results" / "*.csv")))
for i, f in enumerate(files):
    print(i, Path(f).name)
```

```python
untagged = [f for f in files if Path(f).stem.endswith("multilingual-e5-base")]
v2_file = next(f for f in reversed(untagged) if "hybrid" in pd.read_csv(f)["method"].values)
print("v2:", Path(v2_file).name)
print("v3:", Path(files[-1]).name)

old = pd.read_csv(v2_file)
new = pd.read_csv(files[-1])

cols = ["qid", "query", "recall@10", "top_ref"]
o = old[(old.method == "hybrid") & (old.type == "synthetic_quran")][cols]
n = new[(new.method == "hybrid") & (new.type == "synthetic_quran")][["qid", "recall@10", "top_ref"]]
m = o.merge(n, on="qid", suffixes=("_v2", "_v3"))

print("Lebih baik di v3:", (m["recall@10_v3"] > m["recall@10_v2"]).sum())
print("Lebih teruk di v3:", (m["recall@10_v3"] < m["recall@10_v2"]).sum())
m[m["recall@10_v3"] != m["recall@10_v2"]]
```
